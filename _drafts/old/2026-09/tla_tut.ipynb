# Intro

The temporal logic of actions (TLA) is a logic with a community and tooling designed for verification of software systems.

Unlike many other logics, it has an emphasis on time and dynamics.

It is most useful in specifying and checking high level properties involving concurrency.

For fine grained specification of mathematical algorithms where time is of secondary concern, it is not really the best choice, although is does have some capabilities in that direction.

Modern computing is turtles on turtles. The underlying hardware is highly concurrent even for assembly operations that are presented with illusion of serialization to the user.

```python
x = Int('x')
y = Int('y')

s = Solver()
s.add(x > 2)
s.add(y < 10)
s.add(x + 2 *y == 7)

print(s.check())
print(s.model())
```

```python
%%file /tmp/Poly.tla
--------- MODULE Poly ----------

EXTENDS Integers, Sequences
VARIABLES 
    \* @type: Int;
    x,
    \* @type: Int;
    y

Init ==
    /\ x > 2
    /\ y < 10

Next == UNCHANGED <<x,y>>

Prob == x + 2 * y /= 7

============================
```

```python
tla.Module.of_file("/tmp/Poly.tla")
```

```python
import kdrag.solvers.tla as tla

tla.apalache_check("/tmp/Poly.tla", init="Init", inv="Prob")
```
