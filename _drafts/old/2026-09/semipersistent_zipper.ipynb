Semi persistent zippers as proof objects
There are laws that compress up and down moves




```python
from dataclasses import dataclass

@dataclass
class App:
    f : str
    args : list["App"]

@dataclass
class Zipper:
    trail : list[tuple[str, list[App], list[App]]] # ctx
    t : App

```

```python
type Id = int

@dataclass
class Node:
    f : str
    args : list[Id]


@dataclass
class Arena:
    nodes : list[Node]


@dataclass
class Ref:
    arena : Arena
    idx : Id


    
```
