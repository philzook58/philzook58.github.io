```python
from dataclasses import dataclass
from enum import Enum

class Mode(Enum):
    IN = 0
    OUT = 1

class Term: ...
@dataclass
class Var(Term):
    name : str

@dataclass
class App(Term):
    f : str
    args : list[Term]

@dataclass
class Clause:
    head : Term
    modes : list[Mode] # modes of the head arguments
    body : list[Term]

type Prog = list[Clause]

type Env = dict[str, Term]
def interp(query : Term, mode : list[Mode], prog : Prog):
    while query:
        for clause in prog:
            if clause.head.f == query.f and zip(clause.modes, mode):
                    # Interpret the body
                    for term in clause.body:
                        interp(term, [Mode.IN] * len(term.args), prog)
```
