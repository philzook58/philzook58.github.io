```python
%%file /tmp/test.py
import gdb
#help(gdb.Breakpoint)
#gdb.Breakpoint("foo", gdb.BP_WATCHPOINT, gdb.WP_WRITE)
#gdb.Breakpoint("bar", gdb.BP_WATCHPOINT, gdb.WP_WRITE)
#gdb.execute("run")
#gdb.BreakPoint(gdb.Watchpoint, "pc", gdb.BP_WATCHPOINT, gdb.WP_WRITE)
class TraceWatch(gdb.Breakpoint):
    def __init__(self, expression, trace):
        super().__init__(
            expression,
            type=gdb.BP_WATCHPOINT,
            wp_class=gdb.WP_WRITE,
            internal=True,
        )
        self.trace = trace

    def stop(self):
        self.trace.sample()
        return False

class Trace:
    def __init__(self, names):
        self.trace = []
        self.names = names
        self.watchpoints = [
            TraceWatch(name, self) for name in names
        ]
    def sample(self):
        # possibly record pc
        self.trace.append({name: int(gdb.parse_and_eval(name)) for name in self.names})

gdb.execute("break main")
gdb.execute("run")
trace = Trace(["foo", "bar", "$rax"])
gdb.execute("continue")
#while True:
#    trace.event()
#    gdb.execute("continue")
print(trace.trace)
```
