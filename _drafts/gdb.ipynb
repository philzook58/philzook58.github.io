```python
%%file /tmp/fifo.c

#include <stdio.h>
#include <random.h>

#define FIFO_SIZE 8
int len = 0;
int[FIFO_SIZE] fifo;

int put(int v){
    if len >= FIFO_SIZE return -1;
    fifo[len] = v;
    len++;
    return 0;
}

int get(){
    if len <= 0 return -1;
    int v = fifo[0];
    for(int i = 1; i < len; i++){
        fifo[i-1] = fifo[i];
    }
    len--;
    return v;
}


int main() {
    for(;;){
        put(random());
        printf("Got: %d\n", get());
    }
}
```
