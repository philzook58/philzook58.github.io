````python
from dataclasses import dataclass
@dataclass
class TermSeq:
    ctx : int   # dom
    #nports : int # cod
    terms : list[Term]
    def well_formed(self) -> bool: ...
    def dom(self) -> int:
        return self.ctx
    def cod(self) -> int:
        return len(self.terms)
class Term: ...
@dataclass
class Var(Term):
    n : int
@dataclass
class Decl:
    f : str
    arity : int
    nports : int
@dataclass
class App(Term):
    decl : Decl
    out : int
    args : TermSeq

readdecl = Decl("read", 1, 2)
write = Decl("write", 2, 1)
# read : S -> V otimes S
read = TermSeq(1, [  App("read", 0, 2, TermSeq(1, [Var(0)]), App("read", 1, 2, TermSeq(1, [Var(0)])))])
# write : V otimes S -> S
write = TermSeq(2, [  App("write", 0, 1, TermSeq(2, [Var(0), Var(1)]))])

````
