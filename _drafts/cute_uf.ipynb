"So much side reading. @p… I think I've read your post in the past, before I knew you. It's full of ideas and perspectives. Let me offer yet another one: the array programming perspective. In array programming, a far cry from type theory, thinnings are ubiquitous, although they are never called by that name:
Filtering a vector a by a predicate p is a two step process: first apply the predicate to the vector obtaining a bitmask, and then compress the vector by the bitmask: (p a)/a.
Similarly, sorting a vector a is a two step process: first grade the vector to obtain a sorting permutation, and then apply the permutation to the vector: a[⍋a]. (Interestingly, while this is conceptually simpler, it is computationally more expensive.)
Conversion from the bitmask to index representation is done by the indices/where operator ⍸. The operator is invertible, but the inverse is flawed, since the information about the domain is lost.
There's an interesting extension of these ideas from booleans to natural number on the one hand, and from subvectors of indices to such with repetition on the other.
Widenings have also been there from the start. Widening a vector a by a boolean vector b is called expansion, and is notationally symmetric to compression: b\a."
