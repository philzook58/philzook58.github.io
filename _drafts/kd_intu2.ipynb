```python
class Theorem:
    topgoal : Term
    lemmas : list[Term] 
    def qed(self):
        return prove(self.topgoal, by=self.lemmas) 
    def have(self, thm, by=[])
        self.lemmas.append(prove(thm, by=by + self.lemmas))
        return self
```
