````python
@dataclass
class Poly:
    coeffs: list[Term] = field(default_factory=list)
    def __add__(self, other: "Poly") -> "Poly":
        assert isinstance(other, Poly)
        return Poly(sorted(self.coeffs + other.coeffs))
    def __mul__(self, other: "Poly") -> "Poly":
        assert isinstance(other, Poly)
        return Poly(sorted([sorted(a + b) for a in self.coeffs for b in other.coeffs]))
    @staticmethod
    def sing(t : Term) -> Poly:
        assert isinstance(t, Term)
        return Poly([[t]])
    def __repr__(self):
        return f"Poly(" + " + ".join([f"{' '.join(map(str, c))}" for c in self.coeffs]) + ")"
    @staticmethod
    def of_term(t : Term) -> Poly:
        if t.decl == Add:
            return Poly.of_term(t.args[0]) + Poly.of_term(t.args[1])
        elif t.decl == Mul:
            return Poly.of_term(t.args[0]) * Poly.of_term(t.args[1])
        else:
            return Poly.sing(t)
# sorted isn't working because < is being treated as truthy. Ugh
pa = Poly.of_term(a + a * b * a + b * a * a)
pa + pa
````

````python
from dataclasses import dataclass, field
class Sort:
    def thf(self):
        match self:
            case Prop():
                return "$o"
            case Ind():
                return "$i"
            case Arr(dom, cod):
                return f"({dom.thf()} > {cod.thf()})"
class Prop(Sort): ...
class Ind(Sort): ...
@dataclass
class Arr(Sort):
    dom : Sort
    cod : Sort

class Term:
    def thf(self):
        match self:
            case Const(name, sort):
                return name
            case Var(name, sort):
                return name
            case App(f, x):
                return f"({f}@{x})"
            case Bind(kind, vs, body):
                res = f"[]{", ".join(f"{v.name} : {v.sort}" for v in vs)}]: {body.thf()}" 
                match kind:
                    case "forall":
                        return "!" + res
                    case "exists":
                        return "?" + res
                    case "lambda":
                        return "^" + res
                    case _:
                        raise ValueError(f"Unknown binding kind: {kind}")
            case Infix(op, left, right):
                return f"({left} {op} {right})"

@dataclass
class Const(Term):
    name : str
    sort : Sort      
@dataclass
class Var(Term):
    name : str
    sort : Sort
    bound : Term | None = None
    def __post_init__(self):
        assert self.name[0].isupper(), f"Variable names must start with an uppercase letter, got {self.name}"
@dataclass
class App(Term):
    f : Term
    x : Term
@dataclass
class Bind(Term):
    kind : str
    vs : list[Var]
    body : Term
@dataclass
class Infix(Term):
    op : str
    left : Term
    right : Term
def Eq(left, right):
    return Infix("=", left, right)
true = Const("$true", Prop())
false = Const("$false", Prop())
def And(*args):
    if len(args) == 0:
        return true
    elif len(args) == 1:
        return args[0]
    else:
        return Infix("&", args[0], And(*args[1:]))
def Or(*args):
    if len(args) == 0:
        return false
    elif len(args) == 1:
        return args[0]
    else:
        return Infix("|", args[0], Or(*args[1:]))
def Implies(left : Term, right : Term) -> Term:
    return Infix("=>", left, right)
def Not(arg : Term) -> Term:
    return Implies(arg, false)
def ForAll(vs : list[Var], *hyp_concs) -> Term:
    if len(hyp_concs) == 0:
        raise ValueError("ForAll must have at least one hypothesis or conclusion")
    elif len(hyp_concs) == 1:
        body = hyp_concs[0]
    else:
        body = Implies(And(*hyp_concs[:-1]), hyp_concs[-1])
    return Bind("forall", vs, body)
def Exists(vs : list[Var], *hyp_concs) -> Term:
    if len(hyp_concs) == 0:
        raise ValueError("Exists must have at least one hypothesis or conclusion")
    return Bind("exists", vs, And(*hyp_concs))





````

````python
from dataclasses import dataclass
from typing import Optional, Callable
import pprint


@dataclass(frozen=True, slots=True)
class Decl:
    name: str
    arity: int
    bound: Optional["Decl"] = None
    # binds: int
    is_var: bool = False
    # is_prop ?  kind of special. Forall p, p should not be allowed

    def __post_init__(self):
        if self.is_var and not self.name[0].isupper():
            raise ValueError("Variable not capitalized", self)
        elif not self.is_var and self.name[0].isupper():
            raise ValueError("Constant cannot be capitalized")

    def __call__(self, *args: "Term") -> "Term":
        assert (
            len(args) == self.arity
        ), f"Expected {self.arity} arguments, got {len(args)}"
        return Term(self, args)

    @property
    def defn(self) -> "Proof":
        return defns[self]

    @property
    def contract(self) -> "Proof":
        return contracts[self]

    @property
    def is_quantifier(self) -> bool:
        return self is ForAll0 or self is Exists0


defns: dict[Decl, "Proof"] = {}
contracts: dict[Decl, "Proof"] = {}

Set = Decl("set", 1)
Real = Decl("real", 1)

And0 = Decl("and", 2)
Or = Decl("or", 2)
Implies = Decl("imp", 2)
Not = Decl("not", 1)
Iff = Decl("iff", 2)
Eq = Decl("eq", 2)
NEq = Decl("neq", 2)
Add = Decl("add", 2)
Sub = Decl("sub", 2)
Mul = Decl("mul", 2)
Neg = Decl("neg", 1)
LT = Decl("lt", 2)
LE = Decl("le", 2)
ForAll0 = Decl("forall", 2)
Exists0 = Decl("exists", 2)
Call = Decl("call", 2) # app


@dataclass(frozen=True, slots=True)
class Term:
    decl: Decl
    args: tuple["Term", ...]

    def __post_init__(self):
        assert self.decl.is_quantifier or all(isinstance(a, Term) for a in self.args), self

    def eq(self, other: "Term") -> bool:
        assert self.decl.name not in ["exists", "forall"]  # Todo
        return self.decl == other.decl and all(
            s.eq(o) for s, o in zip(self.args, other.args)
        )

    def __add__(self, other: "Term") -> "Term":
        return Add(self, other)

    def __sub__(self, other: "Term") -> "Term":
        return Sub(self, other)

    def __mul__(self, other: "Term") -> "Term":
        return Mul(self, other)

    def __eq__(self, other: "Term") -> "Term":
        return Eq(self, other)

    def __ne__(self, other: "Term") -> "Term":
        return NEq(self, other)

    def __lt__(self, other: "Term") -> "Term":
        return LT(self, other)

    def __neg__(self) -> "Term":
        return Neg(self)

    def __call__(self, *args):
        assert len(args) > 0
        if len(args) == 1:
            return Call(self, args[0])
        else:
            return Call(self, args[0])(*args[1:])

    def __str__(self) -> str:
        match self.decl.name:
            case "and":
                return f"({self.args[0]} & {self.args[1]})"
            case "or":
                return f"({self.args[0]} | {self.args[1]})"
            case "imp":
                return f"({self.args[0]} => {self.args[1]})"
            case "not":
                return f"~{self.args[0]}"
            case "iff":
                return f"({self.args[0]} <=> {self.args[1]})"
            case "eq":
                return f"({self.args[0]} = {self.args[1]})"
            case "neq":
                return f"({self.args[0]} != {self.args[1]})"
            case "forall":
                return (
                    f"![{', '.join(str(arg) for arg in self.args[0])}]: {self.args[1]}"
                )
            case "exists":
                return (
                    f"?[{', '.join(str(arg) for arg in self.args[0])}]: {self.args[1]}"
                )
        if self.decl.arity == 0:
            return self.decl.name
        else:
            return f"{self.decl.name}({', '.join(str(arg) for arg in self.args)})"

    def fvs(self) -> set["Term"]:
        # free variables in the expression
        if self.decl.is_quantifier:
            return self.args[1].fvs() - set(self.args[0])
        elif self.decl.is_var:
            return {self}
        else:
            return set().union(*(arg.fvs() for arg in self.args))

    def decls(self) -> set[Decl]:
        decls = set()
        todo = [self]
        while todo:
            t = todo.pop()
            decl = t.decl
            if decl.is_quantifier:
                todo.append(t.args[1])
            else:
                decls.add(decl)
                todo.extend(t.args)
        return decls


@dataclass(frozen=True, slots=True)
class Proof:
    thm: Term
    reasons: list[object]

    def __repr__(self) -> str:
        return "|- " + str(self.thm)

    def __call__(self, *args: "Term") -> "Proof":
        # instantiate a universally quantified theorem with specific terms
        assert (
            self.thm.decl.name == "forall"
        ), f"Expected forall, got {self.thm.decl.name}"
        assert len(args) == len(
            self.thm.args[0]
        ), f"Expected {len(self.thm.args[0])} arguments, got {len(args)}"
        subst = dict(zip(self.thm.args[0], args))
        new_thm = substitute(self.thm.args[1], subst)
        return Proof(new_thm, self.reasons + [args])


def axiom(p: Term) -> Proof:
    # Ya need axioms. What can I say
    assert (
        len(p.fvs()) == 0
    ), f"Expected closed term, got {p}, free variables: {p.fvs()}"
    return Proof(p, reasons=["axiom"])


import subprocess


def prove0(p: Term, by=[], solver="nanocopi") -> Proof:
    assert isinstance(p, Term), f"Expected Term, got {p}"
    assert (
        len(p.fvs()) == 0
    ), f"Expected closed term, got {p}, free variables: {[str(v) for v in p.fvs()]}"
    with open("/tmp/prob.p", "w") as f:
        for i, b in enumerate(by):
            assert isinstance(b, Proof), f"Expected Proof, got {b}"
            f.write(f"fof(ax{i}, axiom, {b.thm}).\n")
        f.write(f"fof(goal, conjecture, {p}).\n")
    # replace the path with wherever you put nanocopi
    # If you aren't doing intuitionistic logic, use vampire.
    if solver == "nanocopi":
        res = subprocess.run(
            [
                "swipl",
                "-O",
                "-g",
                "assert(prolog(swi)),assert(proof(none)), asserta(logic(intu)), ['/home/philip/Downloads/nanoCoP-i-HT/nanocopi_main.pl'], call_with_time_limit(1,nanocopi_main('/tmp/prob.p',[cut,comp(6)],_)), halt",
            ],
            capture_output=True,
            text=True,
            timeout=2,
        )  # without timeout memory was leaking badly?
        if "is a intu Theorem" in res.stdout:
            return Proof(p, by)
        else:
            vres = vprove0(p, by=by)
            raise ValueError(
                f"Failed to prove {p} with by\n{pprint.pformat(by)}, result: {res.stdout}.\nVampire result: {vres}"
            )
    elif solver == "vampire":
        if all(is_coherent(b) for by in by.thm) and is_coherent(p):
            res = subprocess.run(
                [
                    "vampire",
                    #"--mode",
                    #"casc",
                    "--input_syntax",
                    "tptp",
                    "--time_limit",
                    "1",
                    "--proof",
                    "on",
                    "/tmp/prob.p",
                ],
                capture_output=True,
                text=True,
                timeout=2,
            )
            if "Refutation found" in res.stdout:
                return Proof(p, by)
            else:
                raise ValueError(
                    f"Failed to prove {p} with by\n{pprint.pformat(by)}, result: {res.stdout}"
                )
        else:
            raise ValueError(
                f"Cannot use vampire on non-coherent terms. {[t for t in by if not is_coherent(t.thm)]} and {p if not is_coherent(p) else None}"
            )
    else:
        raise ValueError(f"Unknown solver {solver}")



def And(*args) -> "Term":
    if len(args) == 0:
        return true
    elif len(args) == 1:
        return args[0]
    else:
        return Term(And0, (args[0], And(*args[1:])))


def ForAll(vars: list["Term"], *hyp_conc) -> "Term":
    assert len(hyp_conc) >= 1
    hyp_conc = list(hyp_conc)
    for v in vars:
        assert v.decl.is_var
        if v.decl.bound is not None:
            hyp_conc = [v.decl.bound(v)] + hyp_conc
    if len(hyp_conc) == 1:
        return Term(ForAll0, (tuple(vars), hyp_conc[0]))
    elif len(hyp_conc) == 2:
        return Term(ForAll0, (tuple(vars), Implies(hyp_conc[0], hyp_conc[1])))
    else:
        return Term(ForAll0, (tuple(vars), Implies(And(*hyp_conc[:-1]), hyp_conc[-1])))


def Exists(vars: list["Term"], *hyp_conc) -> "Term":
    assert len(hyp_conc) >= 1
    hyp_conc = list(hyp_conc)
    for v in vars:
        assert v.decl.is_var
        if v.decl.bound is not None:
            hyp_conc = [v.decl.bound(v)] + hyp_conc
    return Term(Exists0, (tuple(vars), And(*hyp_conc)))


def Const(name: str, sort: Decl) -> Term:
    decl = Decl(name, 0)
    t = decl()
    contracts[decl] = axiom(sort(t))
    return t


def Consts(names: str, sort: Decl) -> list[Term]:
    return [Const(name, sort) for name in names.split()]


def Function(name: str, arity: int) -> Decl:
    return Decl(name, arity)


true = Decl("$true", 0)()
false = Decl("$false", 0)()


def Vars(names: str, bound=None) -> list[Term]:
    return [Term(Decl(name, 0, is_var=True, bound=bound), ()) for name in names.split()]


def define(name: str, args: list[Term], body, sort=None):
    f = Function(name, len(args))
    defn = Proof(ForAll(args, f(*args) == body), reasons=["define"])
    global defns
    defns[f] = defn
    return f


from functools import singledispatch


@singledispatch
def lit(x) -> Term:
    raise ValueError("Cannot support x", x)


@lit.register
def _(x: bool) -> Term:
    return true if x else false


# Would it be nice to use binary / decimal lists?
@lit.register
def _(x: int):
    return Const("d" + str(x))


# set?
@lit.register
def _(x: list) -> Term:
    if len(x) == 0:
        return Const("nil")
    else:
        Function("cons", 2)(lit(x[0]), lit(x[1:]))


def prove(thm, by=[], **kwargs):
    cs = [contracts[decl] for decl in thm.decls() if decl in contracts]
    return prove0(thm, by=by + cs, **kwargs)


from dataclasses import field


@dataclass
class Theorem:
    topgoal: Term
    vs : list[Term] = field(default_factory=list)
    hyps : list[Term] = field(default_factory=list)
    lemmas: list[Term] = field(default_factory=list)

    def qed(self):
        return prove(self.topgoal, by=self.lemmas)

    def fixes(self, *vs):
        self.vs.extend(vs)
        return self
    
    def assumes(self, *hyps):
        self.hyps.extend(hyps)
        return self

    def wrap(self, thm):
        # wrap the theorem with the assumptions and variables
        if self.hyps:
            thm = Implies(And(*self.hyps), thm)
        if self.vs:
            thm = ForAll(self.vs, thm)
        return thm

    def have(self, thm, by=[], **kwargs):
        self.lemmas.append(prove(self.wrap(thm), by=by, **kwargs))
        self.hyps.append(thm)
        return self
````
